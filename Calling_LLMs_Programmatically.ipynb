```python
import os
from google.colab import userdata
from getpass import getpass

try:
    # Try to get it from Colab Secrets
    os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI_API_KEY')
    print("OpenAI API key successfully loaded from Colab Secrets.")
except Exception as e:
    print(f"Could not load secret via userdata: {e}")
    # Fallback to manual entry if Secrets are inaccessible
    api_key = getpass("Please enter your OpenAI API key manually: ")
    if api_key:
        os.environ["OPENAI_API_KEY"] = api_key
        print("OpenAI API key set manually for this session.")
    else:
        print("No key entered.")
```
