```python
import os
from dotenv import load_dotenv

load_dotenv(".env")

try:
    userdata = __import__("google.colab", fromlist=["userdata"]).userdata
except ModuleNotFoundError:
    userdata = None

if userdata is not None:
    try:
        colab_api_key = userdata.get("OPENAI_API_KEY")
    except Exception as exc:
        print(f"Could not load the key from Colab Secrets: {exc}")
    else:
        if colab_api_key:
            os.environ["OPENAI_API_KEY"] = colab_api_key
```

```python
from getpass import getpass

if not os.environ.get("OPENAI_API_KEY"):
    api_key = getpass("Please enter your OpenAI API key: ")
    if api_key:
        os.environ["OPENAI_API_KEY"] = api_key
    else:
        print("No key entered.")
```

```python
%pip install -r requirements.txt -q

from openai import OpenAI

client = OpenAI()

prompt = "Explain in simple terms how a large language model generates a response."
response = client.responses.create(
    model="gpt-4o-mini",
    input=prompt,
)

print(f"Prompt: {prompt}")
print("LLM response:")
print(response.output_text)
```
